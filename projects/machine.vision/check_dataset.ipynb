```python
from pathlib import Path
from PIL import Image

ROOT = Path.cwd()
DATA = ROOT / "dataset" / "Alphabet_VISION_Dataset"

for split in ["train", "valid"]:

    folder = DATA / split / "mnist"

    images = sorted(folder.glob("*.png"))

    print(f"\nChecking {split}")
    print("Images:", len(images))

    missing = 0
    invalid = 0

    for path in images:

        txt = path.with_suffix(".txt")

        if not txt.exists():
            missing += 1
            continue

        label = txt.read_text(encoding="utf-8").strip()

        if label not in [str(i) for i in range(10)]:
            invalid += 1

        with Image.open(path) as im:
            if im.size != (28, 28):
                invalid += 1

    print("Missing captions:", missing)
    print("Invalid samples:", invalid)

print("\nValidation completed.")
```
