### 🔴 Tokenization

```python
base_path = "/data/parquet/"

files = [
    "Alphabet_NLP_Dataset.parquet",
    "Alphabet_Codex_Dataset.parquet",
    "Alphabet_Jurisprudence_Dataset.parquet",
    "Alphabet_Quran_Dataset.parquet",
    "Alphabet_Science_Dataset.parquet",
    "Alphabet_Technology_Dataset.parquet",
    "Alphabet_Philologia_Dataset.parquet"
]

datasets = [pd.read_parquet(base_path + file) for file in files]
dataset = pd.concat(datasets, ignore_index=True)

texts = []

for _, row in dataset.iterrows():
    texts.append(str(row["input_text"]).replace("\u200c", "<ZWNJ>"))
    texts.append(str(row["target_text"]).replace("\u200c", "<ZWNJ>"))

tokenizer = Tokenizer(models.BPE(unk_token="|<unk>|"))

tokenizer.pre_tokenizer = pre_tokenizers.ByteLevel(
    add_prefix_space=True
)

trainer = trainers.BpeTrainer(
    vocab_size=10_000,
    special_tokens=["|<unk>|", "<|endoftext|>"],
    min_frequency=2,
)

tokenizer.train_from_iterator(texts, trainer)

tokenizer.decoder = decoders.ByteLevel(
    add_prefix_space=True,
    clean_up_tokenization_spaces=False
)

total_tokens = 0

for text in texts:
    total_tokens += len(tokenizer.encode(text).ids)

json_out = "/data/tokenizer/bpe-tokenizer_alphabet_nlp_dataset_10,000.json"

os.makedirs("/data/tokenizer", exist_ok=True)
tokenizer.save(json_out)

print("🎉 Tokenizer training complete!")
print(f"🔢 tokens: {total_tokens:,}")
print(f"🔹 Vocabulary size: {tokenizer.get_vocab_size():,} tokens")
```

### 🔴 Measuring RAM Usage After Tokenization

```python
tokenizer = Tokenizer.from_file("/data/tokenizer/bpe-tokenizer_alphabet_nlp_dataset_10,000.json")

# Calculate RAM usage after loading
process = psutil.Process(os.getpid())
mem_info = process.memory_info()
ram_mb = mem_info.rss / (1024 ** 2)

print(f"💾 RAM Usage after loading tokenizer: {ram_mb:.2f} MB")
```
