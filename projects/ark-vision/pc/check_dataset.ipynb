```python
from pathlib import Path
import os
import psutil

ROOT = Path.cwd() / "dataset" / "Alphabet_VISION_Dataset"
SPLITS = ("train", "valid", "test")
DATASETS = ("gender_face", "mnist_grayscale", "various_images")

IMAGE_EXTENSIONS = {
    ".png", ".jpg", ".jpeg", ".bmp", ".gif",
    ".tif", ".tiff", ".webp", ".jfif", ".ico"
}

process = psutil.Process(os.getpid())
stats = {
    split: {"images": 0, "missing_captions": 0}
    for split in SPLITS
}

for split in SPLITS:
    for dataset in DATASETS:
        folder = ROOT / split / dataset
        print(f"\nChecking {split}/{dataset}\n{'-' * 40}")

        if not folder.is_dir():
            print(f"ERROR: Folder not found: {folder}")
            continue

        images = sorted(
            p for p in folder.iterdir()
            if p.is_file() and p.suffix.lower() in IMAGE_EXTENSIONS
        )
        missing = sum(not p.with_suffix(".txt").is_file() for p in images)

        stats[split]["images"] += len(images)
        stats[split]["missing_captions"] += missing

        ram = process.memory_info().rss / 1024**2
        print(f"Total images: {len(images)}")
        print(f"Missing captions: {missing}")
        print(f"RAM usage: {ram:.2f} MB")

print("\n" + "=" * 45 + "\nFINAL DATASET SUMMARY\n" + "=" * 45)

for split, s in stats.items():
    print(
        f"\n[{split.upper()}]\n"
        f"Total images: {s['images']}\n"
        f"Missing captions: {s['missing_captions']}"
    )

total_images = sum(s["images"] for s in stats.values())
total_missing = sum(s["missing_captions"] for s in stats.values())

print(
    f"\n[ALL SPLITS]\n"
    f"Total images: {total_images}\n"
    f"Missing captions: {total_missing}"
    f"\n\nFinal RAM usage: {process.memory_info().rss / 1024**2:.2f} MB"
    "\nValidation completed."
)
```
