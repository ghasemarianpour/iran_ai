```python
from pathlib import Path
import psutil
import os

# ==========================================
# Configuration
# ==========================================

ROOT = Path.cwd()
DATA = ROOT / "dataset" / "Alphabet_VISION_Dataset"

SPLITS = ["train", "valid"]
DATASET_NAMES = ["mnist"]

process = psutil.Process(os.getpid())

# ==========================================
# Statistics
# ==========================================

overall = {
    "images": 0,
    "missing_captions": 0,
}

split_stats = {
    split: {
        "images": 0,
        "missing_captions": 0,
    }
    for split in SPLITS
}

# ==========================================
# Dataset checking
# ==========================================

for split in SPLITS:

    for dataset_name in DATASET_NAMES:

        folder = DATA / split / dataset_name

        print(f"\nChecking {split}/{dataset_name}")
        print("-" * 40)

        if not folder.exists():
            print(f"ERROR: Folder not found: {folder}")
            continue

        images = sorted(folder.glob("*.png"))

        image_count = len(images)
        missing = 0

        for path in images:

            txt = path.with_suffix(".txt")

            if not txt.exists():
                missing += 1

        # Update split statistics
        split_stats[split]["images"] += image_count
        split_stats[split]["missing_captions"] += missing

        # Update overall statistics
        overall["images"] += image_count
        overall["missing_captions"] += missing

        # Split report
        print(f"Total images: {image_count}")
        print(f"Missing captions: {missing}")

        ram_mb = process.memory_info().rss / (1024 ** 2)
        print(f"RAM usage: {ram_mb:.2f} MB")

# ==========================================
# Final summary
# ==========================================

print("\n" + "=" * 50)
print("FINAL DATASET SUMMARY")
print("=" * 50)

for split in SPLITS:

    stats = split_stats[split]

    print(f"\n[{split.upper()}]")
    print(f"Total images: {stats['images']}")
    print(f"Missing captions: {stats['missing_captions']}")

# Combined statistics
print("\n" + "-" * 50)
print("[TRAIN + VALID]")
print(f"Total images: {overall['images']}")
print(f"Missing captions: {overall['missing_captions']}")

# ==========================================
# Final RAM usage
# ==========================================

ram_mb = process.memory_info().rss / (1024 ** 2)

print("\n" + "=" * 50)
print(f"Final RAM usage: {ram_mb:.2f} MB")
print("Validation completed.")
print("=" * 50)
```
